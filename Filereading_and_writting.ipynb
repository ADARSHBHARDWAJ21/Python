```python
#csv>>comma seperated files
data = [["Name", "course", "Fee"],
        ["Ajay", "DS", "20000"],
        ["Rahul" , "DA", "30000"]]
```

```python
for i in data:
  print(i)
```

```python
import csv
with open("example_csv.csv" , "w") as f:
  w = csv.writer(f)
  for i in data:
    w.writerow(i)
```

```python
#core python works with reading an csv filee
with open("example_csv.csv" , 'r') as f:
  r = csv.reader(f)
  for i in r:
    print(i)
```

```python
#buffered reading and writting
#it allows  the program to read and write data in chunks
```

```python
import io
with open('test_buf.txt' , "wb") as f:
 file = io.BufferedWriter(f)
 file.write(b"")
 file.write(b"Hello world, how are you?")
 file.write(b"I am studying ds")
 file.flush()
```

```python
with open("test_buf.txt" , "rb") as f:
  file = io.BufferedReader(f)
  data = file.read(100) #read the first chunk of 100 bytes
  print(data)
```
