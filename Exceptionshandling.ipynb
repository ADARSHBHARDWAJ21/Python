```python
#finally block  this block willl be executed aleays
#enven if you enter exception block or not , finally will be executed alwys
#similar to running other block of code
try:
     f =  open("example.txt" , "w")
     f.write("Hello to exceptional handling")
except Exception as e:
     print("There is an issue with the code and the issue is: ", e)
else:
   f.close()
   print(6+3)
   print(" this The file is opened and now I am performing some steps")
finally:
  print("This will be executed always")
```

```python
#custom exception class
class ValidateSalary(Exception):
  def __init__(self , message):
    self.message = message
```

```python
def validate_salary(salary):
  if salary <=0:
    raise ValidateSalary('negative salary is not possible')
  elif salary >= 3000000:
    print("Salary is unexpected")
  else:
    print("Salary is valid")
```

```python
try:
  salary = int(input("Enter your salary"))
  validate_salary(salary)
except ValidateSalary as e:
  print(e)
```
