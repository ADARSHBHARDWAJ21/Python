```python
import logging

logging.basicConfig(
    filename='test_new.log',
    level=logging.INFO,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    force=True
)

logging.info()
```

```python
logging.warning("There can be empty list here")
```

```python
logging.error("Some error has happend")
```

```python
logging.critical("Software has stopped running")
```

```python
#debug>> lowest level of logging . to give detailed information of any variable, mssg
## INFO >> This is used to convey that the code is working as expected
#example "Data analysis is completed"
#warning >> used to indicate that something unexpected happend or potential isuue in the code
#error >> this says there is some serious problem with some function
#critical>> Termination of program/code/software

#the above level of logging is executed in the same order >> debug being lowest priority and critical being highest priority
```

```python
logging.debug("This msg is for debugging")
logging.info("This is my info msg")
logging.warning("this is my warning msg")
logging.shutdown()
```
